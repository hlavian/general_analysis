```python
from pathlib import Path
import numpy as np
import flammkuchen as fl
from glob import glob

import matplotlib.pyplot as plt
from random import randint
```

```python
master = Path(r"Z:\Hagar\main\Herz collaboration")
fish_list = list(master.glob("*_f*"))
fish = fish_list[-5]
fish
```

```python
traces = fl.load(fish / 'filtered_traces.h5', '/detr').T
n_neurons, n_t = np.shape(traces)

t = np.arange(0, n_t) / 5
```

```python
t_end = -1

fig, ax = plt.subplots(2, 1, figsize=(8, 6))
ax[0].imshow(traces[:, :t_end], cmap='coolwarm', vmin=-5, vmax=5, extent=[0, 600, 0, 200])
ax[0].axis('off')
for i in range(10):
    ind = randint(0, n_neurons)
    ax[1].plot(t[:t_end], traces[ind, :t_end] + (20*i), c='k')

ax[1].spines['right'].set_visible(False)
ax[1].spines['top'].set_visible(False)
ax[1].set_xlabel('Time (sec)')
ax[1].set_ylabel('Zscore(F)')
```

```python
fig.savefig(fish / 'example_traces.jpg', dpi=300)
fig.savefig(fish / 'example_traces.pdf', dpi=300)
```
