```python
import tifffile as tiff
import matplotlib.pyplot as plt
from pathlib import Path
import numpy as np
```

```python
master = Path(r"Z:\Hagar\main\Herz collaboration")
files = list(master.glob("*_f*"))
fish = files[-3]
fish
```

```python
anatomy = tiff.imread(fish / 'anatomy_suite2p.tif')
```

```python
n_planes = np.shape(anatomy)[0]
```

```python
n_col = 6
fig, ax = plt.subplots(3,n_col, figsize=(12,8))

anatomy_to_plot = [0, 2, 4, 6, 8, 10, 12, 13, 15, 17, 18, 20, 22, 24, 25, 27, 28, 29, 30, 31]

for i in range(0, n_col*3):
    r = i // n_col
    c = np.mod(i, n_col)
    
    plane = anatomy_to_plot[i]
    
    ax[r, c].imshow(np.rot90(anatomy[plane],2), cmap="gray_r", vmin=10, vmax=1500)
    ax[r, c].axis('off')
```

```python
fig.savefig(fish / 'anatomy_planes.jpg', dpi=720)
```
