```python
master = Path(r"Z:\Hagar\development\dendra anatomy 48 hours intervals")
fish_list = list(master.glob("*_f*"))
fish = fish_list[1]
fish
```

```python
anatomy_file = tiff.imread(fish / "anatomy.tif")
```

```python
# getting resolution
try:
    exp =  TwoPExperiment(fish)
    z_res, x_res, y_res = exp.resolution
except:
    z_res, x_res, y_res = (1,1,1)
```

```python
viewer = napari.view_image(anatomy_file, colormap='gray_r', scale=(z_res, x_res, y_res))
```

```python
cell_centers_ipn = viewer.layers[1].data
#cell_centers_ahb = viewer.layers[2].data
```
