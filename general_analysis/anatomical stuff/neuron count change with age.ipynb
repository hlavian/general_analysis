```python
import numpy as np
import tifffile as tiff
from pathlib import Path
import napari
import flammkuchen as fl
import matplotlib.pyplot as plt
from fimpylab.core.twop_experiment import TwoPExperiment
```

```python
master = Path(r"Z:\Hagar\development\dendra anatomy 24 hours intervals")
tmp_fish = "*_f3"
fish_list = list(master.glob(tmp_fish))
fish_list
```

```python
all_fish = ["*_f0", "*_f1", "*_f2", "*_f3", "*_f4"]
num_fish = len(all_fish)
all_fish2 =  ["*_f7", "*_f8", "*_f9"]
num_fish2 = len(all_fish2)

num_days = 5
num_cells = np.zeros((num_fish+num_fish2, num_days))
num_cells_zeroed = np.zeros((num_fish+num_fish2, num_days))
```

```python
for f in range(num_fish):
    tmp_fish = all_fish[f]
    fish_list = list(master.glob(tmp_fish))
    n_sessions = len(fish_list)
    for s in range(n_sessions):
        fish = fish_list[s]
        cell_ipn = fl.load(fish / "manual_cells.h5", "/cells_ipn")
        num_cells[f, s] = len(cell_ipn)
        num_cells_zeroed[f, s] = len(cell_ipn)
    
    

for f in range(num_fish2):
    tmp_fish = all_fish2[f]
    fish_list = list(master.glob(tmp_fish))
    for s in range(2):
        fish = fish_list[s]
        cell_ipn = fl.load(fish / "manual_cells.h5", "/cells_ipn")
        num_cells[f+5, s+3] = len(cell_ipn)
        num_cells_zeroed[f+5, s] = len(cell_ipn)

num_cells[num_cells == 0] = 'nan'
num_cells_zeroed[num_cells == 0] = 'nan'
```

```python
color_list = ['skyblue', 'dodgerblue', 'gold', 'lightgreen', 'salmon', 'orchid', 'aquamarine', 'deeppink']
```

```python
fig, ax = plt.subplots(1,2, figsize=(8,4))

ax[0].set_ylabel('#cells in IPN')

for i in range(2):
    ax[i].set_xlabel('dpf')
    ax[i].spines['right'].set_visible(False)
    ax[i].spines['top'].set_visible(False)


for f in range(num_fish + num_fish2):
    ax[0].scatter([5,6,7,8,9], num_cells[f], c=color_list[f])


#for f in range(num_fish + num_fish2):
#    ax[1].scatter([5,6,7,8,9], num_cells_zeroed[f] - num_cells_zeroed[f, 0], c=color_list[f])
#ax[1].set_ylim(-2, 20)
    
y= np.nanmean(num_cells[:, :3], axis=0)
error = np.nanstd(num_cells[:, :3], axis=0) / (num_fish + num_fish2)
ax[1].fill_between([5,6,7], y-error, y+error, edgecolor='salmon', facecolor='salmon', alpha=0.8)
ax[1].plot([5,6,7], y, c='r')        
ax[1].set_ylim(10, 40)
```
