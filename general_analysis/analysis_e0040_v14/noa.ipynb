```python
num_params = np.shape(df.iloc[0])[0]
print(num_params)
param = df.iloc[:, 1]
```

```python
gender = df['gender'].values
allergies = df['allergies'].values
pets = df['pets_at_home'].values
books = df['book_per_year'].values
```

```python
# multiple comparisons: 
pval_mul_gen = np.zeros((num_params))
pval_mul_allerg = np.zeros((num_params))
pval_mul_pets = np.zeros((num_params))
pval_mul_books = np.zeros((num_params))

gen0 = np.where(gender == 0)
gen1 = np.where(gender == 1)

allergies0 = np.where(allergies == 0)
allergies1 = np.where(allergies == 1)

pets0 = np.where(pets == 0)
pets1 = np.where(pets > 0)

books0 = np.where(books < 3)
books1 = np.where(books > 5)

for i in range(num_params):
    param = np.asarray(df.iloc[:, i])

    pval_mul_gen[i] = ttest_ind(param[gen0], param[gen1])[1]
    pval_mul_allerg[i] = ttest_ind(param[allergies0], param[allergies1])[1]
    pval_mul_pets[i] = ttest_ind(param[pets0], param[pets1])[1]
    pval_mul_books[i] = ttest_ind(param[books0], param[books1])[1]
```

```python
fig, ax = plt.subplots(1,4, figsize=(8,4))
ax[0].scatter(np.arange(num_params), pval_mul_gen)
ax[0].spines['top'].set_visible(False)
ax[0].spines['right'].set_visible(False)
ax[0].set_title('Gender groups')

ax[1].scatter(np.arange(num_params), pval_mul_allerg)
ax[1].spines['top'].set_visible(False)
ax[1].spines['right'].set_visible(False)
ax[1].set_title('Allergy groups')

ax[2].scatter(np.arange(num_params), pval_mul_pets)
ax[2].spines['top'].set_visible(False)
ax[2].spines['right'].set_visible(False)
ax[2].set_title('Pets groups')

ax[3].scatter(np.arange(num_params), pval_mul_books)
ax[3].spines['top'].set_visible(False)
ax[3].spines['right'].set_visible(False)
ax[3].set_title('Books groups')

for i, txt in enumerate(pval_mul_gen):
    txt_show = np.round(txt * 1000) / 1000
    ax[0].annotate(str(txt_show), (i, pval_mul_gen[i]), fontsize=7)
    
for i, txt in enumerate(pval_mul_allerg):
    txt_show = np.round(txt * 1000) / 1000
    ax[1].annotate(str(txt_show), (i, pval_mul_allerg[i]), fontsize=7)
    
for i, txt in enumerate(pval_mul_pets):
    txt_show = np.round(txt * 1000) / 1000
    ax[2].annotate(str(txt_show), (i, pval_mul_pets[i]), fontsize=7)
    
for i, txt in enumerate(pval_mul_books):
    txt_show = np.round(txt * 1000) / 1000
    ax[3].annotate(str(txt_show), (i, pval_mul_books[i]), fontsize=7)
    
plt.subplots_adjust(wspace=0.4, left=0.1, right=0.9)
```

```python
# splitting the data: 
age = df['age'].values
income = df['monthly_income'].values
bmi = df['bmi'].values
iq = df['iq'].values
happiness = df['happiness'].values
params_for_age = [income, bmi, iq, happiness]

cutoff_age = [20, 25, 30, 35, 40, 45, 50, 55, 60, 65]
num_cutoffs = len(cutoff_age)

pval_age_cutoff = np.zeros((4, num_cutoffs))

for i, cutoff in enumerate(cutoff_age):
    young = np.where(age < cutoff)
    old = np.where(age >= cutoff)
    
    for j in range(4):
        param = params_for_age[j]
        pval_age_cutoff[j, i] = ttest_ind(param[young], param[old])[1]
    
```

```python
np.shape(pval_age_cutoff)
```

```python
fig1, ax1 = plt.subplots(1,4, figsize=(10,5))
title_list = ['income', 'bmi', 'iq', 'happiness']

for i in range(len(params_for_age)):    
    ax1[i].scatter(cutoff_age, pval_age_cutoff[i])
    ax1[i].spines['top'].set_visible(False)
    ax1[i].spines['right'].set_visible(False)
    ax1[i].set_title(title_list[i])
    ax1[i].set_xlabel("Age cutoff")

    for j, txt in enumerate(pval_age_cutoff[i]):
        txt_show = np.round(txt * 1000) / 1000
        ax1[i].annotate(str(txt_show), (cutoff_age[j], pval_age_cutoff[i, j]), fontsize=7)
    
ax1[0].set_ylabel("p value")
  

plt.subplots_adjust(wspace=0.5, left=0.1, right=0.9, bottom=0.15)
```
