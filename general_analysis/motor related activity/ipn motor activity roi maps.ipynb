```python
from pathlib import Path
import flammkuchen as fl
import matplotlib.pyplot as plt
import numpy as np
```

```python
master = Path(r"Z:\Hagar\e0075\v04_4x4")
fish_list = list(master.glob("*_ipn*"))
fish = fish_list[0]
print(fish)
```

```python
suite2p_dir = fish / "suite2p"
all_planes = list(suite2p_dir.glob("*00*"))
num_planes = len(all_planes)
print(num_planes)
```

```python
num_col = 3
num_row = 3
fig, axs = plt.subplots(num_row, num_col, figsize=(8,8))
fig.suptitle("Left bias")

for plane in range(num_row*num_col):
    r = plane // num_col
    c = np.mod(plane, num_col)
    try:
        axs[r,c].axis('off')

        crr_plane = all_planes[plane]
        mot_reg = fl.load(crr_plane / "motor_regressors.h5")
        left = np.asarray(mot_reg['left_bias'])

        suite2p_data = fl.load((crr_plane / "data_from_suite2p_unfiltered.h5"))
        rois = suite2p_data['rois_stack']
        rois_left = np.zeros_like(rois, dtype='float64')
        num_rois = np.max(rois)

        for roi in range(num_rois):
            roi_ind = np.where(rois == roi)
            rois_left[roi_ind] = left[roi]

        axs[r,c].imshow(np.rot90(rois_left[0], 2), vmin=-1, vmax=1, cmap="coolwarm")
    except:
        print("Error")
```

```python
file_name = "motor left bias.jpg"
fig.savefig(fish / file_name, dpi=300)
```
