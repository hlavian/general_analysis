```python
from pathlib import Path
import flammkuchen as fl
import numpy as np
import matplotlib.pyplot as plt

from fimpylab.core.lightsheet_experiment import LightsheetExperiment
```

```python
import seaborn as sns
```

```python
master_path =  Path(r"Z:\Hagar\E0040\v31\pre ablation")
fish_list = list(master_path.glob("*_f*"))
path = fish_list[0]
print(path)
```

```python
n_dir = 8
n_fish = len(fish_list)
```

```python
color = plt.cm.twilight_shifted(np.linspace(0.1,0.9,n_dir))
title_list = [r'$\rightarrow$', r'$\searrow$', r'$\downarrow$', r'$\swarrow$', r'$\leftarrow$', r'$\nwarrow$', r'$\uparrow$', r'$\nearrow$']
```

```python
for path in fish_list:
    fig, ax = plt.subplots(2, n_dir, figsize=(10, 4), sharex=True, sharey=True)
    #suite2p_brain = fl.load(path / "data_from_suite2p_cells_aHB.h5")
    #in_brain_idx = suite2p_brain['coords_idx']

    #traces = fl.load(path / "filtered_traces.h5", "/detr")[:, in_brain_idx]
    #suite2p_data = fl.load(path / "data_from_suite2p_cells.h5")
    #coords = suite2p_data['coords'][in_brain_idx]
    #anatomy = suite2p_data['anatomy_stack']

    exp = LightsheetExperiment(path)
    fs = int(exp.fn)
    beh_df = exp.behavior_log
    stim_duration = int(exp['stimulus']['protocol']['E0040_motions_cardinal']['v31_8dir_plus_hd']['moving_duration']) * fs

    #t = np.arange(np.shape(traces)[1]) / fs
    #len_rec, num_traces = np.shape(traces)

    d = fl.load(path / 'reordered_traces_only_aHB_only_rightleft.h5')
    #left_trials_concat = d['concat_reordered_left_tuned_avg']
    #right_trials_concat = d['concat_reordered_right_tuned_avg']
    left_trials = d['reordered_trials_left_tuned']
    right_trials = d['reordered_trials_right_tuned']
    
    #getting the number of tuned neurons per direction 
    n_tuned_right = np.shape(right_trials)[1]
    n_tuned_left = np.shape(left_trials)[1]
    
    avg_left = np.nanmean(left_trials, axis=2)
    avg_right = np.nanmean(right_trials, axis=2)
    
    history_dependent_responses_r = np.zeros((n_dir, n_tuned_right))
    history_dependent_responses_l = np.zeros((n_dir, n_tuned_left))
    
    for direction in range(n_dir):
        tmp_traces = avg_right[direction, :, :stim_duration]
        history_dependent_responses_r[direction] = np.nanmean(tmp_traces, axis=1)
        
        tmp_traces = avg_left[direction, :, :stim_duration]
        history_dependent_responses_l[direction] = np.nanmean(tmp_traces, axis=1)
        
        #print(history_dependent_responses_l[direction])
        ax[0,direction].spines['right'].set_visible(False)
        ax[1,direction].spines['right'].set_visible(False)
        ax[0,direction].spines['top'].set_visible(False)
        ax[1,direction].spines['top'].set_visible(False)
        
        try:
            ax[0,direction].hist(history_dependent_responses_r[direction], 100, density=False, color=color[direction], alpha=0.6)
            ax[1,direction].hist(history_dependent_responses_l[direction], 100, density=False, color=color[direction], alpha=0.6)
            
        except:
            print("stupid fish")
        ax[0,direction].set_title(title_list[direction] + title_list[0])
        ax[1,direction].set_title(title_list[direction] + title_list[4])
            
    ax[0,0].set_ylabel('Right')
    ax[0,0].set_xlim(-4,4)
    ax[1,0].set_ylabel('Left')
    fig.subplots_adjust(left=0.05, right=0.99, hspace=0.5)
    fig.savefig(path / 'avg response distribution.jpg', dpi=300)
```

```python
### for all fish combined: 
history_dependent_responses_r = [[],[],[],[],[],[],[],[],[]]
history_dependent_responses_l = [[],[],[],[],[],[],[],[],[]]

for path in fish_list:
    
    exp = LightsheetExperiment(path)
    fs = int(exp.fn)
    beh_df = exp.behavior_log
    stim_duration = int(exp['stimulus']['protocol']['E0040_motions_cardinal']['v31_8dir_plus_hd']['moving_duration']) * fs

    d = fl.load(path / 'reordered_traces_only_aHB_only_rightleft.h5')
    left_trials = d['reordered_trials_left_tuned']
    right_trials = d['reordered_trials_right_tuned']
    
    #getting the number of tuned neurons per direction 
    n_tuned_right = np.shape(right_trials)[1]
    n_tuned_left = np.shape(left_trials)[1]
    
    avg_left = np.nanmean(left_trials, axis=2)
    avg_right = np.nanmean(right_trials, axis=2)
   
    for direction in range(n_dir):
        tmp_traces = avg_right[direction, :, :stim_duration]
        history_dependent_responses_r[direction] = history_dependent_responses_r[direction] + np.nanmean(tmp_traces, axis=1).tolist()
        
        tmp_traces = avg_left[direction, :, :stim_duration]
        history_dependent_responses_l[direction] = history_dependent_responses_l[direction] + np.nanmean(tmp_traces, axis=1).tolist()
        
        
```

```python
fig1, ax1 = plt.subplots(2, n_dir, figsize=(10, 4), sharex=True, sharey=True)
    
for direction in range(n_dir):
    ax1[0,direction].spines['right'].set_visible(False)
    ax1[1,direction].spines['right'].set_visible(False)
    ax1[0,direction].spines['top'].set_visible(False)
    ax1[1,direction].spines['top'].set_visible(False)
        
    #try:
    ax1[0,direction].hist(history_dependent_responses_r[direction], 100, density=True, color=color[direction], alpha=0.6)
    ax1[1,direction].hist(history_dependent_responses_l[direction], 100, density=True, color=color[direction], alpha=0.6)

    #except:
    #    print("stupid fish")
    ax1[0,direction].set_title(title_list[direction] + title_list[0])
    ax1[1,direction].set_title(title_list[direction] + title_list[4])
            
ax1[0,0].set_ylabel('Right')
ax1[1,0].set_ylabel('Left')
ax1[0,0].set_xlim(-5,5)
fig1.subplots_adjust(left=0.1, right=0.99, hspace=0.5)
```
