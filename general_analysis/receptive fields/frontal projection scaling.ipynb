```python
import math
import matplotlib.pyplot as plt
import numpy as np
```

```python
def calculate_screen_divisions(radius, num_divisions):
    angles = [i * math.pi / num_divisions for i in range(num_divisions + 1)]
    points = [(radius * math.cos(angle), radius * math.sin(angle)) for angle in angles]
    return points

def plot_screen_divisions(points, radius):
    plt.figure(figsize=(10, 5))
    
    # Plot the screen (half-circle)
    theta = np.linspace(0, np.pi, 100)
    x = radius * np.cos(theta)
    y = radius * np.sin(theta)
    plt.plot(x, y, 'b-')
    
    # Plot division points
    x_points, y_points = zip(*points)
    plt.plot(x_points, y_points, 'ro')
    
    # Plot lines from center to division points
    for point in points:
        plt.plot([0, point[0]], [0, point[1]], 'g--')
    
    plt.axis('equal')
    plt.title('Half-circle Screen Divided into 8 Equal Parts')
    plt.xlabel('X')
    plt.ylabel('Y')
    plt.grid(True)
    plt.show()
```

```python
# Example usage
radius = 10  # Arbitrary radius for the screen
num_divisions = 8

points = calculate_screen_divisions(radius, num_divisions)
plot_screen_divisions(points, radius)

# Print the coordinates of the division points
for i, point in enumerate(points):
    print(f"Division {i}: ({point[0]:.2f}, {point[1]:.2f})")
```

```python
def calculate_screen_divisions(screen_radius, projector_distance, num_divisions):
    angles = np.linspace(0, np.pi, num_divisions + 1)
    screen_points = [(screen_radius * np.cos(angle), screen_radius * np.sin(angle)) for angle in angles]
    
    projection_angles = []
    for x, y in screen_points:
        angle = np.arctan2(y, x - projector_distance)
        projection_angles.append(angle)
    
    return screen_points, projection_angles

def plot_projection_setup(screen_points, projection_angles, screen_radius, projector_distance):
    plt.figure(figsize=(12, 6))
    
    # Plot the screen (half-circle)
    theta = np.linspace(0, np.pi, 100)
    x = screen_radius * np.cos(theta)
    y = screen_radius * np.sin(theta)
    plt.plot(x, y, 'b-', label='Screen')
    
    # Plot projector
    plt.plot(projector_distance, 0, 'ro', markersize=10, label='Projector')
    
    # Plot division points and projection lines
    for (x, y), angle in zip(screen_points, projection_angles):
        plt.plot(x, y, 'go', markersize=5)
        line_x = [projector_distance, x]
        line_y = [0, y]
        plt.plot(line_x, line_y, 'r--', alpha=0.5)
    
    plt.axis('equal')
    plt.title('Projection from Outside onto Half-circle Screen')
    plt.xlabel('X')
    plt.ylabel('Y')
    plt.legend()
    plt.grid(True)
    plt.show()
```

```python

# Example usage
screen_radius = 10
projector_distance = 15  # Projector is 15 units away from the center of the screen
num_divisions = 8

screen_points, projection_angles = calculate_screen_divisions(screen_radius, projector_distance, num_divisions)
plot_projection_setup(screen_points, projection_angles, screen_radius, projector_distance)

# Print the angles for each projection division
for i, angle in enumerate(projection_angles):
    print(f"Division {i}: {np.degrees(angle):.2f}°")
```

```python
def calculate_screen_divisions(screen_radius, projector_position, num_divisions):
    angles = np.linspace(0, np.pi, num_divisions + 1)
    screen_points = [(screen_radius * np.cos(angle), screen_radius * np.sin(angle)) for angle in angles]
    
    projection_angles = []
    for x, y in screen_points:
        dx = x - projector_position[0]
        dy = y - projector_position[1]
        angle = np.arctan2(dy, dx)
        projection_angles.append(angle)
    
    return screen_points, projection_angles

def plot_projection_setup(screen_points, projection_angles, screen_radius, projector_position):
    plt.figure(figsize=(12, 12))
    
    # Plot the screen (half-circle)
    theta = np.linspace(0, np.pi, 100)
    x = screen_radius * np.cos(theta)
    y = screen_radius * np.sin(theta)
    plt.plot(x, y, 'b-', label='Screen')
    
    # Plot projector
    plt.plot(projector_position[0], projector_position[1], 'ro', markersize=10, label='Projector')
    
    # Plot division points and projection lines
    for (x, y), angle in zip(screen_points, projection_angles):
        plt.plot(x, y, 'go', markersize=5)
        line_x = [projector_position[0], x]
        line_y = [projector_position[1], y]
        plt.plot(line_x, line_y, 'r--', alpha=0.5)
    
    plt.axis('equal')
    plt.title('Projection onto Half-circle Screen')
    plt.xlabel('X')
    plt.ylabel('Y')
    plt.legend()
    plt.grid(True)
    
    # Set axis limits to ensure projector is always visible
    max_dim = max(screen_radius, abs(projector_position[0]), abs(projector_position[1]))
    plt.xlim(-max_dim - 1, max_dim + 1)
    plt.ylim(-max_dim - 1, max_dim + 1)
    
    plt.show()
```

```python

# Example usage
screen_radius = 10
num_divisions = 8

# Function to run the calculation and plotting
def run_projection(projector_y):
    projector_position = (0, projector_y)  # Projector on y-axis
    screen_points, projection_angles = calculate_screen_divisions(screen_radius, projector_position, num_divisions)
    plot_projection_setup(screen_points, projection_angles, screen_radius, projector_position)
    
    print(f"Projector position: (0, {projector_y})")
    for i, angle in enumerate(projection_angles):
        print(f"Division {i}: {np.degrees(angle):.2f}°")
    print("\n")

# Run examples with different projector positions
run_projection(15)  # Projector above the screen
run_projection(-15)  # Projector below the screen
run_projection(0)   # Projector at the center of the screen

# You can run this function with any y value to test different projector positions
```
